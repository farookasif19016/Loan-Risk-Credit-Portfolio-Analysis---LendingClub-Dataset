```python
import pandas as pd

df = pd.read_csv(r"C:\Users\samsu\Downloads\loan.csv")
df.dtypes
df.head(10)
df.describe()
```

```python
missing = df.isnull().sum()
missing = missing[missing > 0].sort_values(ascending = False)
missing
```

```python
cols_to_drop = df.columns[df.isnull().sum() == len(df)].tolist()
df = df.drop(columns = cols_to_drop)
print(f"dropped columns {len(cols_to_drop)} in list")
print(df.shape)
```

```python
df = df.drop(columns = ["policy_code", "next_pymnt_d", "url", "desc", "zip_code", "collection_recovery_fee"])
df.shape
```

```python
df.to_csv(r"D:\CV\lending_club_cleaned.csv", index=False)
```

```python
df["revol_util"] = df["revol_util"].str.replace("%", "").astype(float)
```

```python
# replaceing droping values for nulls 

df = df.dropna(subset=["last_credit_pull_d"])
df["title"] = df["title"].fillna("Not Provided")
df["tax_liens"] = df["tax_liens"].fillna(0)
df["revol_util"] = df["revol_util"].fillna(df["revol_util"].median())
df["chargeoff_within_12_mths"] = df["chargeoff_within_12_mths"].fillna(df["chargeoff_within_12_mths"].median())
df["collections_12_mths_ex_med"] = df["collections_12_mths_ex_med"].fillna(df["collections_12_mths_ex_med"].median())
df["last_pymnt_d"] = df["last_pymnt_d"].fillna(df["last_pymnt_d"].mode()[0])
df["pub_rec_bankruptcies"] = df["pub_rec_bankruptcies"].fillna(df["pub_rec_bankruptcies"].median())
df["emp_length"] = df["emp_length"].fillna(df["emp_length"].mode()[0])
df["emp_title"] = df["emp_title"].fillna("Not Provided")
df = df.drop(columns=["mths_since_last_record", "mths_since_last_delinq"])
```

```python
# Data Validation 

print(df["revol_util"].dtype)
print(df["int_rate"].dtype)
```

```python
df["int_rate"] = df["int_rate"].str.replace("%","").astype("float")
print(df["int_rate"].dtype)
```

```python
print(df.isnull().sum().sum())
```

```python
df.to_csv(r"D:\CV\lending_club_cleaned_2.csv", index=False)
```

```python
import pandas as pd
df = pd.read_csv(r"D:\Phase 2 Progress\Data Analytics\lending_club_validated.csv")
df.head(6)
df.tail(10)
df.describe()
df.info()
df.shape

df["loan_status"].value_counts()
```

```python
# Filter the outcome data

df_default = df[df["loan_status"].isin(["Fully Paid","Charged Off"])]
df_default["loan_status"].value_counts()
```

```python
# default rate

default_rate = (df_default["loan_status"] == "Charged Off").mean() * 100

default_rate
```

```python
# Default rate by loan grade

df_default.groupby("grade")["loan_status"].value_counts()
```

```python
# Calculate default rate by grade

grade_default_rate = df_default.groupby("grade")["loan_status"].apply( lambda x: (x == "Charged Off").mean() * 100 )

grade_default_rate.round(2)
```

```python
# Checking Homeownership 

df_default["home_ownership"].value_counts()

df_default.groupby("home_ownership")["loan_status"].value_counts()

df_default.groupby("home_ownership")["loan_status"].apply(lambda x : (x == "Charged Off").mean() * 100).round(2)
```

```python
df_default["purpose"].value_counts()

df_default.groupby("purpose")["loan_status"].apply( lambda x: (x == "Charged Off").mean() * 100).round(2).sort_values(ascending = False)

df_default.groupby("purpose")["loan_status"].value_counts()

df_default.groupby("purpose")["int_rate"].mean().round(2).sort_values(ascending=False)
```

```python
df_default.groupby("loan_status")["annual_inc"].agg(["count", "mean", "median"]).round(2)
```

```python
df_default.groupby("verification_status")["loan_status"].value_counts()

df_default.groupby("verification_status")["loan_status"].apply(
    lambda x: (x == "Charged Off").mean() * 100
).round(2).sort_values(ascending=False)
```

```python
df_default.groupby(
    ["grade", "verification_status"]
)["loan_status"].apply(
    lambda x: (x == "Charged Off").mean() * 100
).round(2)
```
